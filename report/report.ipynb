<h2>1.2.3 <b>Exact Solution</b></h2>
</div>

We have the equation
$$
\begin{align*}
u(t, x, y) &= e^{\imath(k_x x + k_y y - \omega t)} \\
&= \cos(k_x x + k_y y - \omega t) + \imath\sin(k_x x + k_y y - \omega t) \\
&= A + \imath B,
\end{align*}
$$

which makes it so we can write out the wave equation as
$$
\begin{align*}
\frac{\partial^2 u}{\partial t^2} &= c^2(\frac{\partial^2 u}{\partial x^2} + \frac{\partial^2 u}{\partial y^2}) \\
\frac{\partial^2}{\partial t^2}(A+\imath B) &= c^2 \frac{\partial^2}{\partial x^2}(A+\imath B) + c^2 \frac{\partial^2}{\partial y^2}(A+\imath B).
\end{align*}
$$

Since the wave equation is linear, we can split this into real and imaginary parts separately:
$$
\begin{align*}
\frac{\partial^2}{\partial t^2} A + \imath \frac{\partial^2}{\partial t^2} B &= c^2 \left(\frac{\partial^2}{\partial x^2} A + \frac{\partial^2}{\partial y^2} A\right) + \imath c^2 \left(\frac{\partial^2}{\partial x^2} B + \frac{\partial^2}{\partial y^2} B\right).
\end{align*}
$$

Two complex numbers are equal only if their real parts match and their imaginary parts match, which gives us 2 equations:
$$
\begin{align*}
\frac{\partial^2}{\partial t^2} A &= c^2\left(\frac{\partial^2}{\partial x^2} A + \frac{\partial^2}{\partial y^2} A\right) \\
\frac{\partial^2}{\partial t^2} B &= c^2\left(\frac{\partial^2}{\partial x^2} B + \frac{\partial^2}{\partial y^2} B\right)
\end{align*}
$$

Before solving these equations, it is important to also be aware of the fact that we in task 1.2.1 defined an expression for $\omega$ as a function of $c$, $k_x$ and $k_y$, which is as follows:
$$\omega = \sqrt{c^2(k_x^2+k_y^2)}.$$

Now, we can solve this part by part, starting with $A=\cos(k_x x + k_y y - \omega t)$:
$$
\begin{align*}
\frac{\partial A}{\partial t} &= -\omega\sin(\dots), \qquad \frac{\partial^2 A}{\partial t^2} = -\omega^2\cos(\dots) \\
\frac{\partial A}{\partial x} &= -k_x\sin(\dots), \qquad \frac{\partial^2 A}{\partial x^2} = -k_x^2\cos(\dots) \\
\frac{\partial A}{\partial y} &= -k_y\sin(\dots), \qquad \frac{\partial^2 A}{\partial y^2} = -k_y^2\cos(\dots)
\end{align*}
$$

This assembles to
$$
\begin{align*}
\frac{\partial^2}{\partial t^2} A &= c^2\left(\frac{\partial^2}{\partial x^2} A + \frac{\partial^2}{\partial y^2} A\right) \\
-\omega^2 A &= c^2(-k_x^2 A - k_y^2 A) \\
\omega^2 &= c^2(k_x^2+k_y^2)
\end{align*}
$$

From our earlier defined $\omega$, we see that these two sides are equal. $1=1$. To no surprise, the other part $B$ follows this exact same pattern as well, seeing as $\frac{\partial}{\partial x}\sin(ax)=a\cos(ax)$ $\Longrightarrow$ $\frac{\partial^2}{\partial x^2}a\cos(ax)=-a^2\sin(ax)$.

<h2>1.2.4 <b>Dispersion coefficient</b></h2>
</div>

Assuming $m_x=m_y$ such that $k_x=k_y=k$, we get the discrete equation
$$
\begin{align*}
u_{ij}^n=e^{\imath(kh(i+j)-\tilde\omega n\Delta t)},
\end{align*}
$$

which we want to insert into
$$
\begin{align*}
\frac{u_{ij}^{n+1}-2u_{ij}^n+u_{ij}^{n-1}}{\Delta t^2} = c^2 \left(\frac{u_{i+1,j}^n-2u_{ij}^n+u_{i-1,j}^n}{h^2} + \frac{u_{i,j+1}^n-2u_{ij}^2+u_{i,j-1}^n}{h^2}\right).
\end{align*}
$$

Now, what simplifies this whole process is to look at each of these parts separately first, which gives us the much more compact:
$$
\begin{align*}
u_{ij}^{n+1} = e^{\imath(kh(i+j)-\tilde\omega(n+1)\Delta t)} &= e^{\imath(kh(i+j)-\tilde\omega n\Delta t - \tilde\omega\Delta t)} = u_{ij}^n e^{-\imath\tilde\omega\Delta t} \\
u_{ij}^{n-1} = e^{\imath(kh(i+j)-\tilde\omega(n-1)\Delta t)} &= e^{\imath(kh(i+j)-\tilde\omega n\Delta t + \tilde\omega\Delta t)} = u_{ij}^n e^{\imath\tilde\omega\Delta t} \\
u_{i+1, j}^n = e^{\imath(kh(i+1+j)-\tilde\omega n\Delta t)} &= u_{ij}^n e^{\imath kh} \\
u_{i-1, j}^n &= u_{ij}^n e^{-\imath kh} \\
u_{i, j+1}^n &= u_{ij}^n e^{\imath kh} \\
u_{i, j-1}^n &= u_{ij}^n e^{-\imath kh} \\
\end{align*}
$$

With these simplified expressions, we can much easier insert the discrete equation into our wave equation:
$$
\begin{align*}
\frac{u_{ij}^{n+1}-2u_{ij}^n+u_{ij}^{n-1}}{\Delta t^2} &= c^2\left(\frac{u_{i+1,j}^n-2u_{ij}^n+u_{i-1,j}^n}{h^2} + \frac{u_{i,j+1}^n-2u_{ij}^2+u_{i,j-1}^n}{h^2}\right) \\
\frac{u_{ij}^n\left(e^{-\imath\tilde\omega\Delta t}-2+e^{\imath\tilde\omega\Delta t}\right)}{\Delta t^2} &= c^2u_{ij}^n\left(\frac{e^{\imath kh} - 2 + e^{-\imath kh}}{h^2} + \frac{e^{\imath kh} - 2 + e^{-\imath kh}}{h^2} \right) \\
\frac{e^{-\imath\tilde\omega\Delta t}-2+e^{\imath\tilde\omega\Delta t}}{\Delta t^2} &= 2c^2\frac{e^{\imath kh} - 2 + e^{-\imath kh}}{h^2} \\
\end{align*}
$$

Using the identity $e^{\imath\theta}+e^{-\imath\theta}=2\cos(\theta)$, and the fact that $C=\frac{c\Delta t}{h}$, we get:
$$\cos(\tilde\omega\Delta t)-1 = 2C^2(\cos(kh)-1)$$

We can then use the double-angle formula for cosine, $\cos(2\theta) = 1-2\sin^2(\theta/2)$ $\Longrightarrow$ $\cos(2\theta)-1=-2\sin^2(\theta/2)$:
$$
\begin{align*}
-2\sin^2\left(\frac{\tilde\omega\Delta t}{2}\right) &= -4C^2\sin^2\left(\frac{kh}{2}\right) \\
\sin^2\left(\frac{\tilde\omega\Delta t}{2}\right) &= 2C^2\sin^2\left(\frac{kh}{2}\right) \\
\sin\left(\frac{\tilde\omega\Delta t}{2}\right) &= \sqrt{2}C\sin\left(\frac{kh}{2}\right) \\
\frac{\tilde\omega\Delta t}{2} &= \arcsin\left(\sqrt{2}C\sin\left(\frac{kh}{2}\right)\right) \\
\end{align*}
$$

Setting $C=\frac{1}{\sqrt{2}}$:
$$
\begin{align*}
\frac{\tilde\omega\Delta t}{2} &= \arcsin\left(\sqrt{2}\frac{1}{\sqrt{2}}\sin\left(\frac{kh}{2}\right)\right) \\
\tilde\omega &= \frac{2}{\Delta t}\arcsin\left(\sin\left(\frac{kh}{2}\right)\right) \\
\tilde\omega &= \frac{2}{\Delta t}\frac{kh}{2} \\
\tilde\omega &= \frac{kh}{\Delta t} \\
\tilde\omega &= \frac{kc}{C}=  \frac{kc}{1/\sqrt{2}}=\sqrt{2}kc,\\
\end{align*}
$$

since $C=c\Delta t/h$ $\Longrightarrow$ $h/\Delta t = c/C$. If we compare this to our previously defined expression for the exact dispersion coefficient $\omega$, we see that by setting the same assumption of $m_x=m_y$ $\Longrightarrow$ $k_x=k_y=k$, we get:
$$\omega=\sqrt{c^2(k_x^2+k_y^2)}=\sqrt{2c^2k^2}=\sqrt{2}kc = \tilde\omega.$$

So, for CFL number $C=1/\sqrt{2}$, we get $\tilde\omega=\omega$.
